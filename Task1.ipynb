Task 1: Replace Hardcoded Prompts

```python
basic_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple terms for beginners"
)

print(basic_template.format(topic="Artificial Intelligence"))
```

Task 2: Multi-Input Prompt System

```python
multi_template = PromptTemplate(
    input_variables=["topic", "audience", "tone"],
    template="Explain {topic} for {audience} in a {tone} tone"
)
```

```python
print(
    multi_template.format(
        topic="AI",
        audience="beginners",
        tone="friendly"
    )
)

print(
    multi_template.format(
        topic="Python",
        audience="kids",
        tone="fun"
    )
)

print(
    multi_template.format(
        topic="Deep Learning",
        audience="engineers",
        tone="technical"
    )
)
```

Task 3: Prompt Variations Engine

```python
teaching_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} clearly step by step"
)

interview_template = PromptTemplate(
    input_variables=["topic"],
    template="Ask 3 interview questions about {topic}"
)

story_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} as a story"
)
```

```python
topic = "Machine Learning"

print("Teaching Style:")
print(teaching_template.format(topic=topic))

print("\nInterview Style:")
print(interview_template.format(topic=topic))

print("\nStorytelling Style:")
print(story_template.format(topic=topic))
```

Task 4: ChatPromptTemplate System

```python
chat_template = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful {role}."),
    ("human", "Teach me about {topic}.")
])

roles = ["teacher", "interviewer", "motivator"]

for role in roles:
    print("\n" + role.upper() + " ROLE:")

    messages = chat_template.format_messages(
        role=role,
        topic="Neural Networks"
    )

    for message in messages:
        print(message.content)
```

Task 5: Input Validation Layer

```python
def validate_inputs(audience, tone):

    valid_audience = ["beginner", "intermediate", "expert"]
    valid_tone = ["formal", "casual", "fun"]

    if audience not in valid_audience:
        raise ValueError(
            "Invalid audience. Choose from: beginner, intermediate, expert"
        )

    if tone not in valid_tone:
        raise ValueError(
            "Invalid tone. Choose from: formal, casual, fun"
        )

    return audience, tone
```

Testing with valid inputs

```python
print(validate_inputs("beginner", "fun"))
```

Testing with invalid inputs

```python
try:
    print(validate_inputs("kids", "friendly"))
except ValueError as e:
    print(e)
```

Task 6: Prompt Generator App

```python
def generate_prompt(topic, audience, tone, style):

    audience, tone = validate_inputs(audience, tone)

    templates = {
        "teaching": PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Explain {topic} for {audience} in a {tone} teaching style"
        ),

        "interview": PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Ask 3 interview questions about {topic} for {audience} in a {tone} style"
        ),

        "storytelling": PromptTemplate(
            input_variables=["topic", "audience", "tone"],
            template="Explain {topic} for {audience} in a {tone} storytelling style"
        )
    }

    if style not in templates:
        raise ValueError(
            "Invalid style. Choose from: teaching, interview, storytelling"
        )

    selected_template = templates[style]

    return selected_template.format(
        topic=topic,
        audience=audience,
        tone=tone
    )
print(
    generate_prompt(
        topic="Neural Networks",
        audience="beginner",
        tone="fun",
        style="storytelling"
    )
)
print(
    generate_prompt(
        topic="Python",
        audience="expert",
        tone="formal",
        style="teaching"
    )
)

print(
    generate_prompt(
        topic="Artificial Intelligence",
        audience="intermediate",
        tone="casual",
        style="interview"
    )
)
```

Task 7: Template Reusability Test

```python
reuse_template = PromptTemplate(
    input_variables=["topic"],
    template="Explain {topic} in simple language"
)
topics = [
    "Artificial Intelligence",
    "Python",
    "Cloud Computing",
    "Cyber Security",
    "Data Science"
]

for topic in topics:
    result = reuse_template.format(topic=topic)
    print(result)
```
